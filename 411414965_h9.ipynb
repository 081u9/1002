```python
import random
while True:
    s=random.randint(1,5)
    if s==1:
        print("遇到怪物")
    elif s==2:
        print("找到寶物")
    elif s==3:
        print("陷阱跳過")
        continue
    elif s==4:
        print("補充生命")
    elif s==5:
        print("出口結束")
        break
```
